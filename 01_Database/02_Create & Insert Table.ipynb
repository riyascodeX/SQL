# SYNTAX :

```python
'''SYNTAX :
 CREATE TABLE table_name (
    column_name DATA_TYPE,
    column_name DATA_TYPE,
    column_name DATA_TYPE );'''
```

# 1. CREATE TABLE 

```python

%%sql
DROP TABLE IF EXISTS students;
CREATE TABLE students (
    student_id INT,
    student_name VARCHAR(100),
    age INT,
    city VARCHAR(50)
);
```

# 1.1 VIEW THE COLUMN & DATA TYPES & KEYS

```python
%%sql
DESCRIBE students;
```

# SYNTAX :

```python
'''    SYSNTAX :
    -- Insert a single row into specific columns
    INSERT INTO table_name (column1, column2, column3) 
    VALUES (value1, value2, value3);

    -- Insert multiple rows at the same time (Bulk Insert)
    INSERT INTO table_name (column1, column2, column3) 
    VALUES 
        (value1, value2, value3),
        (value4, value5, value6),
        (value7, value8, value9);'''
```

# 2. INSERT A DATA SINGLE OR MULTIPLE ROWS

```python
%%sql
INSERT INTO students
(student_id, student_name, age, city)
VALUES
(2, 'Riyas', 21, 'Madurai'),
(3, 'sameer', 23, 'Coimbatore'),
(4, 'Shalihu', 22, 'Trichy');
```

# 3. RETRIVE ALL COLUMN & ROWS RECORDS

```python
%%sql
SELECT * FROM students
```
