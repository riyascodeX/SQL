# SYNTAX :

```python
'''SYNTAX :
CREATE TABLE table_name (
    column_name datatype PRIMARY KEY,
    column_name datatype NOT NULL,
    column_name datatype UNIQUE,
    column_name datatype CHECK (condition),
    column_name datatype DEFAULT value
);

DEFINITION :
 Constraint  | Simple meaning                  
 ------------| -----------------
 PRIMARY KEY | Unique identifier               
 NOT NULL    | Value must be given            
 UNIQUE      | No duplicate values             
 CHECK       | Value must satisfy condition    
 DEFAUL      | Automatic value if not provided 
 FOREIGN KEY | Connects two tables             '''
```

# 1. CREATE CONSTRAINTS TABLE 

```python
# PRIMARY KEY,NOT NULL,UNIQUE,CHECK,DEFAULT --->CONSTRAINTS
%%sql
CREATE TABLE employee_test (
    employee_id INT PRIMARY KEY,
    employee_name VARCHAR(100) NOT NULL,
    email VARCHAR(100) UNIQUE,
    age INT CHECK (age >= 18),
    city VARCHAR(50) DEFAULT 'Chennai'
);
```

# 1.1 INSERT DATA INTO CONSTRAINTS TABLE

```python
%%sql
# Method 1: Specifying both column names and values 
    INSERT INTO employee_test (employee_id, employee_name, email, age, city) 
        VALUES (1, 'Arun Kumar', 'arun@example.com', 25, 'Chennai');


# Method 2: Inserting data without providing column names Order must match table schema
    INSERT INTO employee_test
        VALUES (2, 'Priya Sharma', 'priya@example.com', 22, 'Bangalore');


# Method 3: Inserting multiple rows in a single query
    INSERT INTO employee_test (employee_id, employee_name, email, age, city)
        VALUES 
        (3, 'Vijay Raj', 'vijay@example.com', 30, 'Madurai'),
        (4, 'Deepa Ram', 'deepa@example.com', 28, 'Coimbatore');


# Method 4: Relying on the DEFAULT constraint Omitting 'city' automatically sets it to 'Chennai'
    INSERT INTO employee_test (employee_id, employee_name, email, age)
        VALUES (5, 'Suresh Kumar', 'suresh@example.com', 24);
```
