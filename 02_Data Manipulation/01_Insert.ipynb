# CREATE TABLE

```python
%%sql
CREATE TABLE insert_students (
    student_id INT,
    student_name VARCHAR(100),
    age INT
);
```

# SYNTAX :

```python
'''INSERT INTO table_name (column1, column2, column3)
    VALUES (value1, value2, value3);'''
```

# 1. INSERT SINGLE ROWS

```python
%%sql
INSERT INTO insert_students (student_id, student_name, age)
VALUES (1, 'Arun', 22);
```

```python
%%sql
select * from insert_students;
```
