# CREATE TABLE

```python
%%sql
CREATE TABLE students_update (
    student_id INT,
    student_name VARCHAR(100),
    age INT,
    city VARCHAR(50)
);

# INSERT DATA
INSERT INTO students_update
(student_id, student_name, age, city)
VALUES
(1, 'Arun', 22, 'Chennai'),
(2, 'Priya', 21, 'Madurai'),
(3, 'Rahul', 23, 'Coimbatore');

# CHECH DATA
SELECT * FROM students_update
```

# SYNTAX :

```python
'''UPDATE table_name SET column_name = new_value
    WHERE condition;'''
```

# 1. UPDATE SINGLE ROWS

```python
%%sql
UPDATE students_update SET age = 24
    WHERE student_id = 1;
```

# 1.1 CHECK UPDATED DATA

```python
%%sql
SELECT * FROM students_update;
```

# 2. UPDATE MULTIPLE ROWS

```python
%%sql
UPDATE students_update SET age = 25,city = 'Bangalore'
    WHERE student_id = 2;
```

# 2.1 CHECK UPDATED DATA

```python
%%sql
SELECT * FROM students_update;
```
