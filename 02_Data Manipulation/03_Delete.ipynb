# CREATE TABLE

```python
%%sql
CREATE TABLE students_delete (
    student_id INT,
    student_name VARCHAR(100),
    age INT,
    city VARCHAR(50)
);

# INSERT DATA
INSERT INTO students_delete
(student_id, student_name, age, city)
VALUES
(1, 'Arun', 22, 'Chennai'),
(2, 'Priya', 21, 'Madurai'),
(3, 'Rahul', 23, 'Coimbatore');

# CHECK DATA
SELECT * FROM students_delete;
```
