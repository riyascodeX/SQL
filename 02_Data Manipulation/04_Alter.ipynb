# CREATE TABLE

```python
%%sql
CREATE TABLE students (
    student_id INT,
    student_name VARCHAR(100),
    age INT
);

select * from students;
```

# CHECK DATA BEFORE ALTER

```python
%%sql
ALTER TABLE students
ADD city VARCHAR(50);
```

# 1.1 CHECK ADDED COLUMN

```python
%%sql
select * from students
```

# 2. MODIFY COLUMN

```python
%%sql
ALTER TABLE students
MODIFY student_name VARCHAR(150);
```

# 2.1 CHECK MODIFIED COLUMN

```python
%%sql
select * from students
```

# 3. RENAME COLUMN

```python
%%sql
ALTER TABLE students
RENAME COLUMN city TO location;
```

```python
%%sql
ALTER TABLE students
DROP COLUMN age;
```
