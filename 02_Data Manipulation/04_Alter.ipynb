# CREATE TABLE

```python
%%sql
CREATE TABLE students (
    student_id INT,
    student_name VARCHAR(100),
    age INT
);

select * from students;
```

# CHECK DATA BEFORE ALTER

# 1. ADD Column

```python
%%sql
ALTER TABLE students
ADD city VARCHAR(50);

# CHECK DATA
select * from students
```

# 2. MODIFY COLUMN

```python
%%sql
ALTER TABLE students
MODIFY student_name VARCHAR(150);

# CHECK DATA
select * from students
```

# 3. RENAME COLUMN

```python
%%sql
ALTER TABLE students
RENAME COLUMN city TO location;

# CHECK DATA
select * from students
```

# 4. DROP Column

```python
%%sql
ALTER TABLE students
DROP COLUMN age;

# CHECK DATA
select * from students
```

# 5. RENAME Table

```python
%%sql
ALTER TABLE students
RENAME TO student_details;

# CHECK DATA
select * from student_details;
```
