# CREATE TABLE

```python
%%sql
CREATE TABLE students_select (
    student_id INT,
    student_name VARCHAR(100),
    age INT,
    city VARCHAR(50),
    course VARCHAR(50),
    marks INT
);

# INSERT DATA
INSERT INTO students_select
(student_id, student_name, age, city, course, marks)
VALUES
(1, 'Arun', 22, 'Chennai', 'SQL', 85),
(2, 'Priya', 21, 'Madurai', 'Python', 90),
(3, 'Rahul', 23, 'Chennai', 'SQL', 75),
(4, 'Meena', 22, 'Coimbatore', 'Power BI', 88),
(5, 'Karthik', 24, 'Madurai', 'SQL', 95),
(6, 'Divya', 21, 'Chennai', 'Python', 82),
(7, 'Vijay', 25, 'Trichy', 'Power BI', 70),
(8, 'Anitha', 23, 'Madurai', 'SQL', 90);

# CHECK DATA
SELECT * FROM students_select;
```

# 1. SELECT

```python
%%sql
SELECT student_name, age, city FROM students_select
```

# 2. DISTINCT

```python
%%sql
SELECT DISTINCT(course) FROM students_select;
```

# 3. ALIAS

```python
%%sql
SELECT 
    student_name AS name,
    marks AS score 
FROM students_select;
```

# 4. WHERE

```python
%%sql
SELECT * FROM students_select 
WHERE age = 22;
```

# 5. AND

```python
%%sql
SELECT * FROM students_select 
WHERE city = 'Chennai' AND marks > 80;
```

# 6. OR

```python
%%sql
SELECT * FROM students_select
WHERE city = 'Chennai' OR city = 'Madurai';
```

# 7. NOT

```python
%%sql
SELECT * FROM students_select 
WHERE NOT course = 'SQL';
```

# 8. IN

```python
%%sql
SELECT * FROM students_select
WHERE city IN ('Chennai', 'Madurai');
```

# 9. BETWEEN

```python
%%sql
SELECT * FROM students_select
WHERE marks BETWEEN 80 AND 90;
```

# 10. LIKE

```python
%%sql
SELECT * FROM students_select
WHERE student_name LIKE 'A%';
```

# 11. NULL

```python
%%sql
# RULE : THIS DATASET DOSEN'T NULL SO UPDATE NULL THEN USE NULL CONCEPTS
UPDATE students_select SET city = NULL
WHERE student_id = 8;

SELECT * FROM students_select 
WHERE city IS NULL;
```

# 12. NOT NULL

```python
%%sql
SELECT * FROM students_select
WHERE city IS NOT NULL;
```

# 13. ORDER BY ASC

```python
%%sql
SELECT * FROM students_select
ORDER BY marks ASC;
```

# 14. ORDER BY DESC

```python
%%sql
SELECT * FROM students_select
ORDER BY marks DESC;
```
