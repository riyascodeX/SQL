```python
%%sql
SELECT * FROM students_select;
```

# 1. SUM

```python
%%sql
SELECT
    SUM(marks) AS total_marks
FROM students_select;
```

# 2. AVERAGE

```python
%%sql
SELECT
    AVG(marks) AS average_marks
FROM students_select;
```

# 3. MINIMUM & MAXIMUM

```python
%%sql
SELECT
    MIN(marks) AS minimum_marks,
    MAX(marks) AS maximum_marks
FROM students_select;
```

# 4. COUNT

```python
%%sql
SELECT COUNT(student_name) FROM students_select;
```

# 5. GROUP BY

```python
%%sql
SELECT
    city,
    COUNT(*) AS total_students
FROM students_select
GROUP BY city;
```

# 6. HAVING

```python
%%sql
SELECT
    course,
    AVG(marks) AS average_marks
FROM students_select
GROUP BY course
HAVING AVG(marks) > 80;
```
