```python
'''
-- QUICK MEMORY

INNER JOIN  → Matching rows only
LEFT JOIN   → All LEFT + matching RIGHT
RIGHT JOIN  → All RIGHT + matching LEFT
FULL JOIN   → All rows from BOTH
SELF JOIN   → Same table with itself
CROSS JOIN  → Every possible combination
'''
```

# 1. INNER JOIN

```python
%%sql
SELECT
    c.customer_name,
    o.order_id,
    o.product,
    o.amount
FROM join_customers c
INNER JOIN join_orders o
    ON c.customer_id = o.customer_id;
```

# 2. LEFT JOIN

```python
%%sql
SELECT
    c.customer_name,
    o.order_id,
    o.product,
    o.amount
FROM join_customers c
LEFT JOIN join_orders o
    ON c.customer_id = o.customer_id;
```

# 3. RIGHT JOIN

```python
%%sql
SELECT
    c.customer_name,
    o.order_id,
    o.product,
    o.amount
FROM join_customers c
RIGHT JOIN join_orders o
    ON c.customer_id = o.customer_id;
```

# 4. FULL JOIN

```python
%%sql
#LEFT JOIN + RIGHT JOIN + UNION = FULL OUTER JOIN
SELECT
    c.customer_name,
    o.order_id,
    o.product,
    o.amount
FROM join_customers c
LEFT JOIN join_orders o
    ON c.customer_id = o.customer_id

UNION

SELECT
    c.customer_name,
    o.order_id,
    o.product,
    o.amount
FROM join_customers c
RIGHT JOIN join_orders o
    ON c.customer_id = o.customer_id;
```

# 5.SELF JOIN

```python
%%sql
SELECT
    e.employee_name AS employee,
    m.employee_name AS manager
FROM join_employees e
LEFT JOIN join_employees m
    ON e.manager_id = m.employee_id;
```

# 6. CROSS JOIN

```python
%%sql
SELECT
    c.customer_name,
    o.product
FROM join_customers c
CROSS JOIN join_orders o;
```

```python
'''
============================================================
-- JOINS
============================================================
JOIN is used to combine rows from two or more tables
using a related column.

-- Practice Tables:
join_customers
join_orders
join_employees
-- ============================================================


 ============================================================
-- 1. INNER JOIN
============================================================
--Definition:
    Returns only the rows that have a match in BOTH tables.

-- Memory:
    INNER = MATCHING ROWS ONLY

-- Syntax:
    SELECT columns
    FROM table1
    INNER JOIN table2
        ON table1.column = table2.column;

============================================================
-- 2. LEFT JOIN
============================================================
-- Definition:
    Returns ALL rows from the LEFT table
    and matching rows from the RIGHT table.

-- If there is no match, RIGHT table columns become NULL.

-- Memory:
    LEFT = ALL LEFT + MATCHING RIGHT

-- Syntax:
    SELECT columns
    FROM table1
    LEFT JOIN table2
        ON table1.column = table2.column;

============================================================
-- 3. RIGHT JOIN
============================================================
-- Definition:
    Returns ALL rows from the RIGHT table
    and matching rows from the LEFT table.

-- If there is no match, LEFT table columns become NULL.

-- Memory:
    RIGHT = ALL RIGHT + MATCHING LEFT

-- Syntax:
    SELECT columns
    FROM table1
    RIGHT JOIN table2
        ON table1.column = table2.column;

============================================================
-- 4. FULL JOIN
============================================================
-- Definition:
    Returns ALL rows from BOTH tables.

    MySQL does not directly support FULL OUTER JOIN.
    We can achieve it using LEFT JOIN + RIGHT JOIN + UNION.

-- Memory:
FULL = ALL LEFT + ALL RIGHT
--Syntax:
    LEFT JOIN
    UNION
    RIGHT JOIN

============================================================
-- 5. SELF JOIN
============================================================
-- Definition:
    A table is joined with itself.
    Used when rows in the same table are related to each other.

-- Example:
    Employee → Manager

-- Memory:
    SELF = SAME TABLE

-- Syntax:
    SELECT ...
    FROM table t1
    JOIN table t2
        ON t1.column = t2.column;

============================================================
-- 6. CROSS JOIN
============================================================
-- Definition:
    Combines EVERY row from the first table
    with EVERY row from the second table.
    No matching condition is required.

-- Memory:
    CROSS = EVERY COMBINATION

-- Formula:
    Rows in Table A × Rows in Table B

-- Example:
    5 customers × 5 orders = 25 combinations


-- QUICK MEMORY

INNER JOIN  → Matching rows only
LEFT JOIN   → All LEFT + matching RIGHT
RIGHT JOIN  → All RIGHT + matching LEFT
FULL JOIN   → All rows from BOTH
SELF JOIN   → Same table with itself
CROSS JOIN  → Every possible combination


'''
```
