```python
%%sql
CREATE TABLE function_students (
    student_id INT,
    student_name VARCHAR(100),
    city VARCHAR(50),
    marks INT,
    fee DECIMAL(10,2),
    birth_date DATE
);

#INSERT DATA
INSERT INTO function_students
VALUES
(1, 'Arun Kumar', 'Chennai', 85, 15000.00, '2002-05-10'),
(2, 'Priya S', 'Madurai', 92, 18000.00, '2001-08-15'),
(3, 'Rahul M', 'Coimbatore', 76, 12000.00, '2003-02-20'),
(4, 'Meena R', 'Chennai', 88, 16000.00, '2002-11-05'),
(5, 'Karthik P', 'Madurai', 95, 20000.00, '2001-03-25');

# CHECK DATA
SELECT * FROM function_students;
```

# STRING FUNCTIONS

## 1. UPPER

```python
'''UPPER()  → Text→ UPPERCASE '''
```

```python
%%sql
SELECT UPPER(student_name) AS upper_name
FROM function_students;
```

## 2. LOWER

```python
'''LOWER()  → Text → lowercase'''
```

```python
%%sql
SELECT LOWER(student_name) AS lower_name
FROM function_students;
```

## 3. LENGTH

```python
'''LENGTH() → Count characters'''
```

```python
%%sql
SELECT
    student_name,
    LENGTH(student_name) AS name_length
FROM function_students;
```

## 4. CONCAT

```python
'''CONCAT() → Join text'''
```

```python
%%sql
SELECT
    CONCAT(student_name, ' - ', city) AS student_info
FROM function_students;
```

## 5. SUBSTRING

```python
'''SUBSTRING() → Extract part of text'''
```

```python
%%sql
SELECT
    SUBSTRING(student_name, 1, 5) AS short_name
FROM function_students;
```

## 6. TRIM

```python
'''TRIM() → Remove extra spaces'''
```

```python
%%sql
SELECT TRIM('  SQL Learning  ');
```

# NUMERIC FUNCTIONS

## 6. ROUND

```python
'''ROUND() → Round decimal'''
```

```python
%%sql
SELECT
fee,
ROUND(fee, 0) AS rounded_fee
FROM function_students;
```

## 7. CEIL

```python
'''CEIL() → Round up'''
```

```python
%%sql
SELECT 
fee,
CEIL(fee / 1000) AS result
FROM function_students;
```

## 8. FLOOR

```python
'''FLOOR() → Round down'''
```

```python
%%sql
SELECT
fee,
FLOOR(fee / 1000) AS result
FROM function_students;
```

## 9. ABS

```python
'''ABS() → Negative → positive'''
```

```python
%%sql
SELECT ABS(-100) AS result;
```

## 10. MOD

```python
'''MOD() → Get remainder'''
```

```python
%%sql
SELECT MOD(marks, 10) AS remainder
FROM function_students;
```

# DATE FUNCTIONS

## 11. YEAR

```python
'''YEAR() → Get year from date'''
```

```python
%%sql
SELECT
    student_name,
    birth_date,
    YEAR(birth_date) AS birth_year
FROM function_students;
```

## 12. MONTH

```python
'''MONTH() → Get month from date'''
```

```python
%%sql
SELECT
    student_name,
    birth_date,
    MONTH(birth_date) AS birth_month
FROM function_students;
```

## 13. DAY

```python
'''DAY() → Get day from date'''
```

```python
%%sql
SELECT
    student_name,
    birth_date,
    DAY(birth_date) AS birth_day
FROM function_students;
```

## 14. CURRENT DATE

```python
'''CURDATE() → Get current date'''
```

```python
%%sql
SELECT CURDATE();
```

## 15. DATEDIFF

```python
'''DATEDIFF() → Difference between dates'''
```

```python
%%sql
SELECT
    student_name,
    DATEDIFF(CURDATE(), birth_date) AS days_difference
FROM function_students;
```

## 16. DATE FORMAT

```python
'''DATE_FORMAT() → Change date format'''
```

```python
%%sql
SELECT
    student_name,
    DATE_FORMAT(birth_date, '%d-%m-%Y') AS formatted_date
FROM function_students;
```

# NULL FUNCTIONS

## 17. COALESCE

```python
'''COALESCE() → First non-NULL value'''
```

```python
%%sql
SELECT
    COALESCE(city, 'Unknown') AS city
FROM function_students;
```

## 18. IFNULL

```python
'''IFNULL() → Replace NULL with value'''
```

```python
%%sql
SELECT
    IFNULL(city, 'Unknown') AS city
FROM function_students;
```
