```python
%%sql
CREATE TABLE case_students (
    student_id INT,
    student_name VARCHAR(50),
    marks INT
);

# INSERT DATA
INSERT INTO case_students
VALUES
(1, 'Arun', 85),
(2, 'Priya', 92),
(3, 'Rahul', 65),
(4, 'Meena', 45),
(5, 'Karthik', 30);

# CHECK DATA
SELECT * FROM case_students
```

# 1. CASE WHEN

## SYNTAX :

```python
''' CASE
        WHEN condition THEN result
        WHEN condition THEN result
        ELSE result
    END'''
```

```python
%%sql
SELECT
    student_name,
    marks,
    CASE
        WHEN marks >= 80 THEN 'Excellent'
        WHEN marks >= 60 THEN 'Good'
        WHEN marks >= 40 THEN 'Pass'
        ELSE 'Fail'
    END AS grade
FROM case_students;
```
