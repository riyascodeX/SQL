# 1. SUBQUERY

```python
# SUBQUERY → Query inside another query

%%sql
SELECT
    student_name,
    marks
FROM subquery_students
WHERE marks > (
    SELECT AVG(marks)
    FROM subquery_students
);
```

# 2. SCALAR SUBQUERY

```python
# SCALAR SUBQUERY → Returns one value

%%sql
SELECT
    student_name,
    marks,
    (
        SELECT AVG(marks)
        FROM subquery_students
    ) AS average_marks
FROM subquery_students;
```

# 3. CORRELATED SUBQUERY

```python
# CORRELATED SUBQUERY → Inner query depends on outer query

%%sql
SELECT
    s.student_name,
    s.course,
    s.marks
FROM subquery_students s
WHERE s.marks > (
    SELECT AVG(s2.marks)
    FROM subquery_students s2
    WHERE s2.course = s.course
);
```

# 4. EXISTS

```python
# EXISTS → Checks whether matching row exists

%%sql
SELECT
    s.student_name
FROM subquery_students s
WHERE EXISTS (
    SELECT 1
    FROM subquery_students s2
    WHERE s2.course = 'SQL'
);
```
