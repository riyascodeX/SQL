# CREATE TABLE

```python
%%sql
CREATE TABLE cte_students (
    student_id INT,
    student_name VARCHAR(50),
    course VARCHAR(50),
    marks INT
);

# INSERT DATA
INSERT INTO cte_students
VALUES
(1, 'Arun', 'SQL', 85),
(2, 'Priya', 'Python', 92),
(3, 'Rahul', 'SQL', 65),
(4, 'Meena', 'Power BI', 88),
(5, 'Karthik', 'SQL', 95);

# CHECK DATA
SELECT * FROM cte_students
```

# 1. SINGLE CTE

## SYNTAX :

```python
'''
WITH cte_name AS (
    SELECT ...
)
SELECT *
FROM cte_name;'''
```

```python
%%sql

# WITH → Create CTE
# AS   → Define CTE query
# CTE  → Temporary named result

WITH high_scorers AS (
    SELECT
        student_name,
        marks
    FROM cte_students
    WHERE marks > 80
)

# CHECK CTE
SELECT *
FROM high_scorers;
```

# 2. MULTIPLE CTE

```python
%%sql

# CTE → One temporary result
# Multiple CTE → Multiple temporary results

WITH high_scorers AS (
    SELECT *
    FROM cte_students
    WHERE marks > 80
),
sql_students AS (
    SELECT *
    FROM high_scorers
    WHERE course = 'SQL'
)

# CHECK CTE
SELECT *
FROM sql_students;
```

# 3. CTE WITH AGGREGATION

```python
%%sql
WITH course_average AS (
    SELECT
        course,
        AVG(marks) AS average_marks
    FROM cte_students
    GROUP BY course
)
SELECT *
FROM course_average
WHERE average_marks > 80;
```

# 4. RECURSIVE CTE

```python
%%sql

# RECURSIVE CTE → CTE references itself

WITH RECURSIVE numbers AS (
    SELECT 1 AS number

    UNION ALL

    SELECT number + 1
    FROM numbers
    WHERE number < 5
)
SELECT *
FROM numbers;
```
