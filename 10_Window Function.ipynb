# CONCEPTS : 

```python
'''
SELECT
    column_name,
    window_function() OVER (
        PARTITION BY column_name
        ORDER BY column_name
    ) AS result
FROM table_name;

--BASICS:

OVER()           → Defines the window
PARTITION BY     → Divide into groups
ORDER BY         → Define order inside group 

--RANKING:

ROW_NUMBER   → Unique number
RANK         → Tie + Skip
DENSE_RANK   → Tie + No Skip
PERCENT_RANK → Relative rank
NTILE        → Divide into groups

--VALUE/OFFSET

LAG          → Previous row
LEAD         → Next row
FIRST_VALUE  → First value
LAST_VALUE   → Last value
NTH_VALUE    → Nth value
CUME_DIST    → Cumulative distribution
'''
```

# RANKING FUNCTIONS

## 1.ROW_NUMBER()

```python
%%sql

# ROW_NUMBER → Give unique sequential number

SELECT
    student_name,
    marks,
    ROW_NUMBER() OVER (ORDER BY marks DESC) AS row_num
FROM window_students;
```

## 2. RANK()

```python
%%sql

# RANK → Same value gets same rank, skips next rank

SELECT
    student_name,
    marks,
    RANK() OVER (ORDER BY marks DESC) AS rank_num
FROM window_students;
```

## 3. DENSE_RANK()

```python
%%sql

# DENSE_RANK → Same value gets same rank, no skipping

SELECT
    student_name,
    marks,
    DENSE_RANK() OVER (ORDER BY marks DESC) AS `dense_rank`
FROM window_students;
```

## 4. PERCENT_RANK()

```python
%%sql

# PERCENT_RANK → Shows relative rank as percentage

SELECT
    student_name,
    marks,
    PERCENT_RANK() OVER (ORDER BY marks) AS `percent_rank`
FROM window_students;
```

## 5. NTILE()

```python
%%sql

# NTILE → Divide rows into approximately equal groups

SELECT
    student_name,
    marks,
    NTILE(4) OVER (ORDER BY marks DESC) AS group_num
FROM window_students;
```

# VALUE / OFFSET FUNCTIONS

## 6. LAG()

```python
%%sql

# LAG → Get previous row's value

SELECT
    student_name,
    marks,
    LAG(marks) OVER (ORDER BY marks) AS previous_marks
FROM window_students;
```

## 7. LEAD()

```python
%%sql

# LEAD → Get next row's value

SELECT
    student_name,
    marks,
    LEAD(marks) OVER (ORDER BY marks) AS next_marks
FROM window_students;
```

## 8. FIRST_VALUE()

```python
%%sql

# FIRST_VALUE → Get first value in the window

SELECT
    student_name,
    marks,
    FIRST_VALUE(marks) OVER (
        ORDER BY marks DESC
    ) AS highest_marks
FROM window_students;
```

## 9. LAST_VALUE()

```python
%%sql

# LAST_VALUE → Get last value in the window

SELECT
    student_name,
    marks,
    LAST_VALUE(marks) OVER (
        ORDER BY marks
        ROWS BETWEEN UNBOUNDED PRECEDING
        AND UNBOUNDED FOLLOWING
    ) AS last_marks
FROM window_students;
```

## 10. NTH_VALUE()

```python
%%sql

# NTH_VALUE → Get Nth value from the window

SELECT
    student_name,
    marks,
    NTH_VALUE(marks, 2) OVER (
        ORDER BY marks DESC
        ROWS BETWEEN UNBOUNDED PRECEDING
        AND UNBOUNDED FOLLOWING
    ) AS second_highest
FROM window_students;
```

## 11. CUME_DIST()

```python
%%sql

# CUME_DIST → Percentage of rows at or below current value

SELECT
    student_name,
    marks,
    CUME_DIST() OVER (ORDER BY marks) AS cumulative_distribution
FROM window_students;
```
