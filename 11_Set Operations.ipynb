# CREATE TABLE 1

```python
%%sql

# TABLE 1
CREATE TABLE set_students_1 (
    student_id INT,
    student_name VARCHAR(50),
    city VARCHAR(50)
);

# INSERT DATA
INSERT INTO set_students_1
VALUES
(1, 'Arun', 'Chennai'),
(2, 'Priya', 'Madurai'),
(3, 'Rahul', 'Coimbatore');

# CHECK DATA
SELECT * FROM set_students_1
```

# CREATE TABLE 2

```python
%%sql
CREATE TABLE set_students_2 (
    student_id INT,
    student_name VARCHAR(50),
    city VARCHAR(50)
);

#INSERT DATA
INSERT INTO set_students_2
VALUES
(3, 'Rahul', 'Coimbatore'),
(4, 'Meena', 'Trichy'),
(5, 'Karthik', 'Salem');

# CHECK DATA
SELECT * FROM set_students_2
```

# 1. UNION

```python
%%sql

# UNION      → Combine + Remove duplicates
# UNION ALL  → Combine + Keep duplicates

SELECT
    student_id,
    student_name,
    city
FROM set_students_1

UNION

SELECT
    student_id,
    student_name,
    city
FROM set_students_2;
```

# 2. UNION ALL

```python
%%sql

# UNION      → Combine + Remove duplicates
# UNION ALL  → Combine + Keep duplicates

SELECT
    student_id,
    student_name,
    city
FROM set_students_1

UNION ALL

SELECT
    student_id,
    student_name,
    city
FROM set_students_2;
```
