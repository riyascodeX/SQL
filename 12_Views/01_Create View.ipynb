# CREATE TABLE

```python
%%sql

CREATE TABLE view_students (
    student_id INT,
    student_name VARCHAR(50),
    course VARCHAR(50),
    marks INT
);

# INSERT DATA
INSERT INTO view_students
VALUES
(1, 'Arun', 'SQL', 85),
(2, 'Priya', 'Python', 92),
(3, 'Rahul', 'SQL', 65),
(4, 'Meena', 'Power BI', 88),
(5, 'Karthik', 'SQL', 95);

# CHECK DATA
SELECT * FROM view_students;
```

# SYNTAX :

```python
'''
CREATE VIEW view_name AS
SELECT ...
FROM table_name
WHERE condition;

--BASICS
VIEW :
→ Virtual table
→ Saved SELECT query

CREATE VIEW → Create view
SELECT       → Use view
DROP VIEW    → Delete view
'''
```

# 1. CREATE VIEW

```python
%%sql
CREATE VIEW sql_students AS
SELECT
    student_id,
    student_name,
    marks
FROM view_students
WHERE course = 'SQL';
```

# 1.2 USE VIEW

```python
%%sql
SELECT * FROM sql_students;
```

# 1.3 SHOW VIEW

```python
%%sql
SHOW FULL TABLES
WHERE TABLE_TYPE = 'VIEW';
```

# 1.4 DELETE VIEW

```python
%%sql
DROP VIEW sql_students;
```
