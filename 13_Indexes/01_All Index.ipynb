# CREATE TABLE

```python
%%sql
CREATE TABLE index_students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50),
    city VARCHAR(50),
    marks INT
);

# INSERT DATA
INSERT INTO index_students
VALUES
(1, 'Arun', 'Chennai', 85),
(2, 'Priya', 'Madurai', 92),
(3, 'Rahul', 'Coimbatore', 75),
(4, 'Meena', 'Chennai', 88),
(5, 'Karthik', 'Madurai', 95);

# CHECK DATA
SELECT * FROM index_students
```

# 1. CREATE SINGLE COLUMN INDEX

## SYNTAX :

```python
'''
CREATE INDEX index_name
ON table_name (column_name);'''
```

```python
%%sql
CREATE INDEX idx_city
ON index_students (city);
```

# 1.1 INDEX USAGE

```python
%%sql
SELECT *
FROM index_students
WHERE city = 'Chennai';
```

# 1.3 CHECK INDEX

```python
%%sql
SHOW INDEX
FROM index_students;
```

# 2. CREATE MULTIPLE COLUMN INDEX

```python
%%sql
CREATE INDEX idx_city_marks
ON index_students (city, marks);
```

# 2.1 INDEX USAGE

```python
%%sql
SELECT *
FROM index_students
WHERE city = 'Madurai'
AND marks > 90;
```

# 3. DROP INDEX

## SYNTAX :

```python
'''
DROP INDEX index_name
ON table_name;'''
```

```python
%%sql
DROP INDEX idx_city
ON index_students;
```
