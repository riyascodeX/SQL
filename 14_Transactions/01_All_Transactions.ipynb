# CREATE TABLE

```python
%%sql
CREATE TABLE transaction_students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50),
    marks INT
);

# INSERT DATA
INSERT INTO transaction_students
VALUES
(1, 'Arun', 85),
(2, 'Priya', 92),
(3, 'Rahul', 75);

# CHECK DATA
SELECT * FROM transaction_students
```

# 1. COMMIT

```python
%%sql

# COMMIT → Changes permanently save.

START TRANSACTION;

UPDATE transaction_students
SET marks = 90
WHERE student_id = 1;

COMMIT;
```

# 2. ROLLBACK

```python
%%sql

# ROLLBACK → changes undo use before commit.

START TRANSACTION;

UPDATE transaction_students
SET marks = 50
WHERE student_id = 2;

ROLLBACK;
```

# 3. SAVEPOINT

```python
%%sql
START TRANSACTION;

UPDATE transaction_students
SET marks = 90
WHERE student_id = 1;

SAVEPOINT point1;

UPDATE transaction_students
SET marks = 50
WHERE student_id = 2;

ROLLBACK TO point1;

COMMIT;
```
