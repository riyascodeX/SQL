# CREATE TABLE

```python
%%sql
CREATE TABLE optimization_students (
    student_id INT PRIMARY KEY,
    student_name VARCHAR(50),
    city VARCHAR(50),
    marks INT
);

#INSERT DATA
INSERT INTO optimization_students
VALUES
(1, 'Arun', 'Chennai', 85),
(2, 'Priya', 'Madurai', 92),
(3, 'Rahul', 'Coimbatore', 75),
(4, 'Meena', 'Chennai', 88),
(5, 'Karthik', 'Madurai', 95);

# CHECK DATA
SELECT * FROM optimization_students
```

```python
'''
CONCEPTS :
EXPLAIN
→ Check query execution

INDEX
→ Faster searching

SELECT needed columns
→ Avoid SELECT *

WHERE
→ Filter data

LIMIT
→ Reduce returned rows

Avoid unnecessary functions
→ Better index usage '''
```

# NORMAL QUERY

```python
%%sql
SELECT *
FROM optimization_students
WHERE city = 'Chennai';
```

# 1. QUERY OPTIMIZATION - EXPLAIN

```python
%%sql
EXPLAIN
SELECT *
FROM optimization_students
WHERE city = 'Chennai';
```

# 2. QUERY OPTIMIZATION - AVOID SELECT

```python
%%sql

# Avoid SELECT * → Select only needed columns

SELECT
    student_name,
    marks
FROM optimization_students;
```

# 3.  QUERY OPTIMIZATION - USE INDEX

```python
%%sql
CREATE INDEX idx_city
ON optimization_students(city);
```

```python
%%sql
EXPLAIN
SELECT
    student_name,
    marks
FROM optimization_students
WHERE city = 'Chennai';
```

# 4. QUERY OPTIMIZATION - USE LIMIT

```python
%%sql
SELECT *
FROM optimization_students
ORDER BY marks DESC
LIMIT 3;
```
